```python
import pandas as pd
import numpy as np
```

```python
# Get the AMR profiles from the NG-STAR types

# Load NG-STAR types for all complete genomes
ngstar = pd.read_csv('../../results/resistance_alleles/ngstar_complete_genomes.tsv', sep = '\t')
ngstar['NG-STAR'].replace(to_replace = '-', value = 'NA', inplace = True)
ngstar['strain'] = ngstar['strain'].str.split('.fa', expand = True)[0]
ngstar.rename({'strain':'wgs_id'}, inplace = True, axis = 'columns')

# Load NG-STAR AMR profiles
ngstar_profiles = pd.read_csv('../../resources/ngstar_profiles_metadata_2026-03-28T11_44_57-05_00.csv', encoding='unicode_escape')
ngstar_profiles.rename({'Sequence Type':'NG-STAR'}, inplace = True, axis = 'columns')
ngstar_profiles['NG-STAR'] = ngstar_profiles['NG-STAR'].astype('str')
```

```python
# Merge types and profiles
resistance_alleles = ngstar.merge(ngstar_profiles, on = 'NG-STAR', how = 'left')
resistance_alleles = resistance_alleles[['wgs_id', 'NG-STAR', 'penA', 'mtrR', 'porB', 'ponA', 'gyrA', 'parC', '23S', 'AMR Markers']]
```

```python
# Save results
resistance_alleles.to_csv('../../results/resistance_alleles/complete_genomes_resistance_profiles.csv')
```

```python
# Get resistance MICs where they exist

# Load MIC data
metadata = pd.read_csv("../../input_data/lab_strains_hybrid_genomes_metadata.csv", index_col = 0)
metadata = metadata[['wgs_id', 'has_mics', 'azithromycin', 
                     'cefixime', 'ceftriaxone', 'ciprofloxacin',
                     'gentamicin', 'penicillin', 'spectinomycin', 
                     'tetracycline', 'zoliflodacin']]

# Merge with resistance alleles
resistance_alleles_with_mic = resistance_alleles.merge(metadata, on = 'wgs_id', how = 'left')
resistance_alleles_with_mic['has_mics'].fillna('N', inplace = True)

# Save results
resistance_alleles_with_mic.to_csv('../../results/resistance_alleles/complete_genomes_resistance_profiles_with_mics.csv')
```
