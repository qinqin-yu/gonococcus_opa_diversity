```python
import pandas as pd
import numpy as np
```

```python
# Read in opa metadata
df = pd.read_csv('../../results/opa_metadata_locus.csv', index_col = 0)

# Get the opas that did not fall in any LCB
opa_no_lcb = df[pd.isnull(df['ref_lcb_start'])][['strain', 'id', 'start', 'stop', 'strand']]
opa_no_lcb.reset_index(inplace = True, drop = True)
opa_no_lcb.to_csv('../../results/progressivemauve/opa_not_in_lcb.csv')
```

```python
# Get the opas that were not assigned a locus and save to file
opa_locus_other = df[pd.isnull(df['locus'])][['strain', 'id', 'start', 'stop', 'strand', 'start_reordered_flipped', 'strand_flipped']]
opa_locus_other.reset_index(inplace = True, drop = True)
opa_locus_other.rename({'start_reordered_flipped':'start_FA1090', 'strand_flipped':'strand_FA1090'}, inplace = True, axis = 1)
opa_locus_other.to_csv('../../results/progressivemauve/opa_locus_other.csv')
```
